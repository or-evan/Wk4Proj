```python
from pathlib import Path
from datetime import datetime
import pandas as pd

def load_oni_data(path="data/oni.ascii.txt"):
    season_month = {
        "DJF": 1, "JFM": 2, "FMA": 3, "MAM": 4,
        "AMJ": 5, "MJJ": 6, "JJA": 7, "JAS": 8,
        "ASO": 9, "SON": 10, "OND": 11, "NDJ": 12
    }

    rows = []
    for line in Path(path).read_text().splitlines():
        fields = line.split()

        if len(fields) < 4 or fields[0] not in season_month:
            continue

        try:
            season = fields[0]
            year = int(fields[1])
            total = float(fields[2])
            anom = float(fields[3])
        except ValueError:
            continue

        if anom > 2:
            category = "VSE"
        elif anom > 1.5:
            category = "SE"
        elif anom > 1:
            category = "ME"
        elif anom > 0.5:
            category = "WE"
        elif anom > -0.5:
            category = "N"
        elif anom > -1:
            category = "WL"
        elif anom > -1.5:
            category = "ML"
        elif anom > -2:
            category = "SL"
        else:
            category = "VSL"

        rows.append({
            "season": season,
            "year": year,
            "date": datetime(year, season_month[season], 1),
            "total": total,
            "anom": anom,
            "category": category
        })

    return pd.DataFrame(rows).sort_values("date").reset_index(drop=True)


oni = load_oni_data()
oni.head()
```

```python
oni.to_csv("data/oni.csv", index=False)
```
